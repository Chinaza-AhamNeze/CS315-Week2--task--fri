```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time
import re

options = Options()
driver = webdriver.Chrome(options=options)

print("Opening page...")
driver.get("https://mydramalist.com/shows/top")

print("Waiting 10 seconds — watch the browser window now")
time.sleep(10)

print("Grabbing page source...")
html = driver.page_source
driver.quit()

soup = BeautifulSoup(html, "html.parser")
titles = soup.find_all("h6")
print(f"Found {len(titles)} <h6> tags")

for h6 in titles[:20]:
    link = h6.find("a")
    if not link:
        continue
    title = link.get_text(strip=True)
    drama_url = link["href"]
    container = h6.find_parent()
    text = container.get_text(" ", strip=True)
    info_match = re.search(r"([A-Za-z ]+(?:Drama|Movie|Special)) - (\d{4}), (\d+) episodes?", text)
    rating_match = re.search(r"\b(\d\.\d)\b", text)
    print(title)
    print(" URL:", drama_url)
    if info_match:
        print(" Info:", info_match.group(0))
    if rating_match:
        print(" Rating:", rating_match.group(1))
    print()
```
